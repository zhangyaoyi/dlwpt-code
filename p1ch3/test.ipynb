```python
import torch

# 创建一个随机张量来模拟我们的图像数据
images = torch.randn(1, 3, 4, 4)
print("Original shape:", images.shape)
```

```python
images
```

```python
images_t1 = images.transpose(1, 2)
print("Transposed shape:", images_t1.shape)
images_t1
```

```python
images_t2 = images.transpose(1, 3)
```

```python
print("Transposed shape:", images_t2.shape)
```
